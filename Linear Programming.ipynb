```python
from pulp import *
import pandas as pd 
import numpy as np
```

```python
n_factory=2
n_customers=8
n_products=3
```

```python
#Cost martrix for production
#cost_matrix_production=np.array([[1.5,1.5,1.56],[1.56,1.57,1.55]])
cost_matrix_production=np.array([[1.5,1.56],[1.5,1.57],[1.56,1.55]])
```

```python
#Cost matrix for shipping 
cost_matrix_shipping=np.array([[0.273,0.317],[0.4908,0.2598],[0.4402,0.0926],[0.515,0.372],[0.2268,0.3368],[0.1696,0.3368],[0.1476,0.372],[0.1586,0.4754]])
```

```python
#price matrix for product 
price_matrix_product=np.array([[12.57,12.57,12.72,11.52,11.13,11.13,11.13,11.13],[12.33,12.33,12.45,11.28,10.92,10.92,10.92,10.92],[12.84,12.84,12.96,11.79,11.4,11.4,11.4,11.4]])
```

```python
#Demand Matrix

cust_demand=np.array([[42900,318830,143600,19350,9870,102970,57490,183400],[25800,57090,31980,27050,34320,71130,175030,295230],[133160,165930,182080,77070,7830,155230,224750,127360]])
```

```python
#Capacity Matrix
factory_capacity=np.array([1200000,1500000])
```

```python
#Price matrix
Selling_price=np.array([[12.57,12.57,12.72,11.52,11.13,11.13,11.13,11.13],[12.33,12.33,12.45,11.28,10.92,10.92,10.92,10.92],[12.84,12.84,12.96,11.79,11.40,11.40,11.40,11.40]])
```

```python
#Duty matrix 
#Duty_cost=np.array([[0.220,0.300,0.090,0.120,0.000,0.000,0.000,0.000],[0.220, 0.300,0.000,0.120,0.042,0.042,0.042,0.042]])
Duty_Cost=np.array([[0.220,0.220]
,[0.300,0.300]
,[0.090,0.000]
,[0.120,0.120]
,[0.000,0.0420]
,[0.000,0.0420]
,[0.000,0.0420]
,[0.000,0.0420]])
```

```python
variable_names=[str(k)+str(j)+str(i) for j in range(1,n_customers+1) for i in range(1,n_factory+1) for k in range(1,n_products+1)]
variable_names.sort()
print("Variable Indices:", variable_names)
```

```python
DV_variables =LpVariable.matrix('X',variable_names,cat='Integer',lowBound=0)
allocation = np.array(DV_variables).reshape(3,8,2)
print('Decision Variable/Allocation Matrix:')
print(allocation)
```

```python
#cost for production 
obj_func = 0

for j in range(0,n_customers):
    obj_func += lpSum(allocation[:,j,:]*cost_matrix_production)

#obj_func=lpSum(allocation*cost_matrix_production_austin)
print(obj_func)
```

```python
#cost for shipping 

for k in range(0,n_products):
    obj_func += lpSum(allocation[k,:,:]*cost_matrix_shipping)

print(obj_func)
```

```python
#price for product
for i in range(0,n_factory):
    obj_func+= lpSum(allocation[:,:,i]*Selling_price)
print (obj_func)

```

```python
#Duty for each cutomer 
for k in range(0,n_factory):
    obj_func+= lpSum(allocation[k,:,:]*Duty_Cost)
print (obj_func)
```

```python
model=LpProblem('Supply-Demand-Problem', LpMinimize)
model+=obj_func
```

```python
#supply constraints
for i in range(n_factory):
    print (lpSum(allocation[k][j][i] for j in range(0,n_customers)  for k in range (0,n_products)) <= factory_capacity[i])
    model+=lpSum(allocation[k][j][i] for j in range(0,n_customers) for k in range (0,n_products)) <= factory_capacity[i],'supply constrains' +str(i)
```

```python
cust_demand
```

```python
#demand constraints
for k in range (0,n_products):
    for j in range(0,n_customers):
        print (lpSum(allocation[k][j][i] for i in range (0,n_factory)) >= cust_demand[k][j])
        model+=lpSum(allocation[k][j][i] for i in range (0,n_factory)) >= cust_demand[k][j],'Demand Constraints'+str(j)+str(k)
```

```python
model
```

```python
model.solve(PULP_CBC_CMD())
status=LpStatus[model.status]
```
